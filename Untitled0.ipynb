```python
# Librerías
# ==============
import networkx as nx
import pandas as pd
import warnings
import matplotlib.pyplot as plt

warnings.filterwarnings ("ignore")
```

```python
# Creación de una instancia de tipo "Grafo" # ============
G = nx.Graph()
```

```python
# Añadir un único nodo
# ===========
fig, ax = plt.subplots( figsize = ( 2, 2 ) )
plt.title( "Grafo con un único nodo" )
G.add_node( "A" )
nx.draw( G, with_labels = True, ax = ax )
print( G )
```

```python
# Añadir multiples nodos
# ===================
lista_nodos = [ "B", "C", "D", "E" ]
G.add_nodes_from( lista_nodos )
```

```python
# Creación de una instancia de tipo "Grafo" # ============
G = nx.Graph()

# Añadir un único nodo
# ===========
G.add_node( "A" )

# Añadir multiples nodos
# ===================
lista_nodos = [ "B", "C", "D", "E" ]
G.add_nodes_from( lista_nodos )

fig, ax = plt.subplots( figsize = ( 3, 3 ) )
plt.title( "Grafo con múltiples nodos" )
nx.draw( G, with_labels = True, ax = ax )
ax.set_xlim([ 1.2*x for x in ax.get_xlim() ])
ax.set_ylim([ 1.2*y for y in ax.get_ylim() ])
print(G)
```
